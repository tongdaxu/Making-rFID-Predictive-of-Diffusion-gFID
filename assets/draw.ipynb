```python
import matplotlib.pyplot as plt
import numpy as np

# 1. The Data
data = {
    "SiT-XL gFID w/o cfg": {
        "SD-VAE": 25.91, "IN-VAE": 25.65, "FLUX1": 34.06, "QW-VAE": 23.76,
        "SD3-VAE": 26.38, "EQ-VAE": 20.81, "VA-VAE": 8.57, "VA-VAE(c64)": 15.4,
        "SOFT-VQ": 15.88, "MAE-TOK": 6.27, "DE-TOK": 11.97, "DM-VAE": 4.65,
        "REPAE-VAE": 12.95, "RAE": 4.25, "PIXEL": 104.29, "JIT": 114.26, "FLUX2": 10.18,
        "PIXGEN": 64.68,
        "PIXEL+REPA": 64.52,
        # "LDM-Conv30": 25.4, "LDM-Conv50": 21.31, 
        "VTP-L": 5.50,
        # "FLUX2+Res": 49.83,
        # "VTPL+Res": 70.21,
        # "LLDM-VTPL-Lossy": 10.42, 
        "Lossless LDM (Ours)": 11.66, 
        # "LLDM-VTPL": 13.70,
    },
    "PSNR": {
        "SD-VAE": 25.68, "EQ-VAE": 24.92, "VA-VAE": 26.31,
        "VA-VAE(c64)": 29.11, "SOFT-VQ": 22.94, "MAE-TOK": 23.52, "FLUX1": 31.09,
        "IN-VAE": 27.44, "DE-TOK": 24.02, "DM-VAE": 21.67, "SD3-VAE": 29.57,
        "QW-VAE": 30.27, "REPAE-VAE": 24.82, "RAE": 18.82, "FLUX2": 34.33,
        # "LDM-Conv30": 31.7, "LDM-Conv50": 33.58,
        "VTP-L": 26.85,
        "PIXEL": np.inf, "JIT": np.inf, "PIXGEN": np.inf, "LLDM-Conv30": np.inf,
        # "FLUX2+Res": np.inf,
        # "VTPL+Res": np.inf,
        # "LLDM-VTPL-Lossy": 33.4, 
        # "LLDM-VTPL": np.inf, 
        "Lossless LDM (Ours)": np.inf, 
    },
}

# 2. Separate Finite and Infinite Data
finite_x, finite_y, finite_labels = [], [], []
inf_y, inf_labels = [], []

for model, gfid in data["SiT-XL gFID w/o cfg"].items():
    if model in data["PSNR"]:
        psnr = data["PSNR"][model]
        if np.isinf(psnr):
            inf_y.append(gfid)
            inf_labels.append(model)
        else:
            finite_x.append(psnr)
            finite_y.append(gfid)
            finite_labels.append(model)

# 3. Create the figure with two subplots (Left for numbers, Right for Infinity)
# width_ratios: 5 parts for finite data, 1 part for infinity
fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True, facecolor='w', figsize=(9, 4.5),
                               gridspec_kw={'width_ratios': [5,2]})
fig.subplots_adjust(wspace=0.08) # Small gap between plots

# 4. Plot Finite Data (ax1)
p1 = ax1.scatter(finite_x, finite_y, color='royalblue', s=80, 
                 edgecolors='white', linewidth=0.5, alpha=0.9, label='Finite PSNR')
for i, txt in enumerate(finite_labels):
    ax1.annotate(txt, (finite_x[i], finite_y[i]), xytext=(5, 5), 
                 textcoords='offset points')

# 5. Plot Infinite Data (ax2)
# We use a dummy X value (0) for all infinite points
p2 = ax2.scatter([0]*len(inf_y), inf_y, color='crimson', marker='X', 
                 s=100, edgecolors='white', linewidth=0.5, label='PSNR = $\infty$')
for i, txt in enumerate(inf_labels):
    ax2.annotate(txt, (0, inf_y[i]), xytext=(10, 0), 
                 textcoords='offset points', va='center')

# 6. Set Axis Limits and Ticks
ax1.set_xlim(min(finite_x)-1, max(finite_x)+1)
ax2.set_xlim(-0.5, 2.5) # Extra space on right for labels
ax2.set_xticks([0])
ax2.set_xticklabels([r'$\infty$'])

# 7. Hide spines and clean up ticks to create the "break" effect
ax1.spines['right'].set_visible(False)
ax2.spines['left'].set_visible(False)
ax2.tick_params(left=False) # Hide ticks on the left of ax2

# 8. Add the "Break" marks (diagonal slashes)
d = .015 # size of the diagonal lines
kwargs = dict(transform=ax1.transAxes, color='k', clip_on=False, linewidth=1)
ax1.plot((1-d, 1+d), (-d, +d), **kwargs) # bottom-right break
ax1.plot((1-d, 1+d), (1-d, 1+d), **kwargs) # top-right break
ax1.set_yscale('log')
# Adjusting break marks for the right axis (multiplier handles different width_ratio)
kwargs.update(transform=ax2.transAxes) 
ax2.plot((-d*5, +d*5), (-d, +d), **kwargs) # bottom-left break
ax2.plot((-d*5, +d*5), (1-d, 1+d), **kwargs) # top-left break
ax2.set_yscale('log')
# 9. Legend (Combining handles from both axes)
handles = [p1, p2]
labels = [h.get_label() for h in handles]
ax1.legend(handles, labels, loc='upper left', frameon=True, shadow=True)

# 10. Final Labels and Styling
ax1.set_ylabel("SiT-XL gFID w/o cfg")

fig.text(0.45, 0.02, 'PSNR (dB)', ha='center')

ax1.grid(True, linestyle=':', alpha=0.6)
ax2.grid(True, axis='y', linestyle=':', alpha=0.6)

plt.show()
```
